```python
from sympy import *
from sympy.plotting import plot
```

```python
x = Symbol('x')
b = 1
```

```python
b = 1
p1 = plot(b * ln(1 + exp(b * x)), show=True, xlim=(-1.5, 1.5), ylim=(0, 5))
```

```python
b = 1.3
p2 = plot(b * ln(1 + exp(b * x)), show=True, xlim=(-1.5, 1.5), ylim=(0, 5))
p1.extend(p2)
```

```python
b = 1.5
p2 = plot(b * ln(1 + exp(b * x)), show=True, xlim=(-1.5, 1.5), ylim=(0, 5))
p1.extend(p2)
```

```python
b = 2
p2 = plot(b * ln(1 + exp(b * x)), show=True, xlim=(-1.5, 1.5), ylim=(0, 5))
p1.extend(p2)
```

```python
p1.show()
```
