```python
import numpy as np
from collections import OrderedDict
import torch
```

```python
def load_npz(file, obj, remove_aux=False):
    state_dict = OrderedDict()
    with np.load(file) as f:
        num_keys = 0
        num_params = 0
        for key in f.keys():
            num_params += sum(f[key].shape)
            num_keys += 1
            print(f"key: {key}              {f[key].shape} {sum(f[key].shape)}")
            if remove_aux and '_aux' in key:
                continue
            names = key.split('/')
            if names[-1] == 'W':
                names[-1] = 'weight'
            elif names[-1] == 'b':
                names[-1] = 'bias'
            elif names[-1] == 'avg_mean':
                names[-1] = 'running_mean'
            elif names[-1] == 'avg_var':
                names[-1] = 'running_var'
            elif names[-1] == 'gamma':
                names[-1] = 'weight'
            elif names[-1] == 'beta':
                names[-1] = 'bias'
            elif names[-1] == 'N':
                names[-1] = 'num_batches_tracked'
            state_dict['.'.join(names)] = torch.tensor(f[key])
    obj.load_state_dict(state_dict)
    print(f"num_keys  : {num_keys}")
    print(f"num_params: {num_params}")
```

```python
from dlshogi.policy_value_network_resnet10_swish import *

model = PolicyValueNetwork()
load_npz('../model/model-2018', model)
```
