```python
sq = np.arange(81)
fsq = sq[hflip_squares]
fsq
```

```python
fsq2 = fsq[hflip_squares]
fsq2
```

```python
sq == fsq2
```

```python
# flipped move labels (directions + hand pieces)

# move direction
hflip_direction = [
    UP, UP_RIGHT, UP_LEFT, RIGHT, LEFT, DOWN, DOWN_RIGHT, DOWN_LEFT, UP2_RIGHT, UP2_LEFT,
    UP_PROMOTE, UP_RIGHT_PROMOTE, UP_LEFT_PROMOTE, RIGHT_PROMOTE, LEFT_PROMOTE, DOWN_PROMOTE, DOWN_RIGHT_PROMOTE, DOWN_LEFT_PROMOTE, UP2_RIGHT_PROMOTE, UP2_LEFT_PROMOTE
]

assert(len(hflip_direction) == len(MOVE_DIRECTION))
assert(len(hflip_direction) == 20)

hflip_direction = np.array(hflip_direction + list(range(20, 27)))
hflip_direction
```

```python
d = np.arange(20)
fd = d[hflip_direction]
fd
```

```python
fd2 = fd[hflip_direction]
fd2
```

```python
d == fd2
```
